# Slide figures

Loads committed comparison and confusion-matrix plots from `results/`.

```python
from pathlib import Path
from IPython.display import Image, display
import pandas as pd

root = Path("..").resolve()
cmp = root / "results/comparison/comparison.csv"
if cmp.exists():
    display(pd.read_csv(cmp))
    display(Image(root / "results/comparison/comparison.png"))
for name in ["logistic_regression", "svm", "random_forest", "xgboost"]:
    png = root / "results" / name / "confusion_matrix.png"
    if png.exists():
        print(name)
        display(Image(png))
```
