# EDA — lake bloom severity

Uses `data/processed/clean_data.csv`. Training features exclude `year`, `distance_to_water_m`, `region`, and `data_provider`.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

root = Path("..").resolve()
df = pd.read_csv(root / "data/processed/clean_data.csv")
features = [
    "tmax_c", "tmin_c", "prcp_mm", "wind_ms", "rain_sum_7d_mm", "tmax_mean_7d_c",
    "B02", "B03", "B04", "B05", "B08", "NDCI", "NDVI", "NDWI",
    "n_water_px", "s2_days_offset", "s2_cloud_pct", "month",
]
print(df.shape)
print(df["severity4"].value_counts().sort_index())
df["severity4"].value_counts().sort_index().plot(kind="bar", title="severity4 counts")
plt.show()
df[features].describe().T
```
